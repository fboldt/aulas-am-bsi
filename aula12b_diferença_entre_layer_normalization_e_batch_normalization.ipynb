# Diferença entre "Layer Normalization" e "Batch Normalization"

```python
import numpy as np

batches_of_sequences = np.array([
    [[1, 2, 3, 4],
     [8, 7, 6, 5],
     [9, 1, 3, 5]],

    [[10, 20, 30, 40],
     [80, 70, 60, 50],
     [90, 10, 30, 50]],
])

print("Original Batch Sequences:")
print(batches_of_sequences)
print("shape:", batches_of_sequences.shape)
print()
```

```python
def layer_normalization(batch_of_sequences):
  mean = batch_of_sequences.mean(axis=-1, keepdims=True)
  std = batch_of_sequences.std(axis=-1, keepdims=True)
  return (batch_of_sequences - mean) / std

normalized_by_layer = layer_normalization(batches_of_sequences)
print("Normalized by Layer:")
print(normalized_by_layer)
print("shape:", normalized_by_layer.shape)
print()
```

```python
print(normalized_by_layer.sum(axis=-1))
```

```python
def batch_normalization(batch_of_sequences):
  mean = batch_of_sequences.mean(axis=-2, keepdims=True)
  std = batch_of_sequences.std(axis=-2, keepdims=True)
  std = std + 1e-8
  return (batch_of_sequences - mean) / std

normalized_by_batch = np.array([batch_normalization(seq) for seq in batches_of_sequences])
print("Normalized by Batch:")
print(normalized_by_batch)
print("shape:", normalized_by_batch.shape)
print()
```

```python
mean = normalized_by_batch.mean(axis=-2)
std = normalized_by_batch.std(axis=-2)

print(mean)
print(std)
```
