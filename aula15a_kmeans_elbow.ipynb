```python
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True)
```

```python
from sklearn.cluster import KMeans

kmeans_k3 = KMeans(n_clusters=3).fit(X)
print(f"kmeans_k3.inertia_ = {kmeans_k3.inertia_}")

kmeans_k8 = KMeans(n_clusters=8).fit(X)
print(f"kmeans_k8.inertia_ = {kmeans_k8.inertia_}")
```

```python
import matplotlib.pyplot as plt

def plot_elbow(X, max_k=9):
  inertias = []
  ks = range(2, max_k+1)
  for k in ks:
    kmeans = KMeans(n_clusters=k).fit(X)
    inertias.append(kmeans.inertia_)
  plt.plot(ks, inertias, marker="o")
  plt.xlabel("Number of clusters (k)")
  plt.ylabel("Inertia")
  plt.title("Elbow Method for Optimal k")
  plt.xticks(ks)
  plt.grid()
  plt.show()

plot_elbow(X, max_k=9)
```

```python
from sklearn.datasets import load_wine

X, y = load_wine(return_X_y=True)
plot_elbow(X, max_k=9)
```

```python
from sklearn.datasets import load_digits
X, y = load_digits(return_X_y=True)
plt.figure(figsize=(10, 6))
plot_elbow(X, max_k=35)
```
