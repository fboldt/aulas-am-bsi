```python
import numpy as np
from matplotlib import pyplot as plt

def create_dataset(n=20, coefs=[1, 1], intercept=-1):
  X = np.random.rand(n, 2)
  y = X @ coefs + intercept
  y = np.sign(y)
  return X, y

def plot_dataset(X, y):
  plt.scatter(X[:,0], X[:,1], c=y, cmap="coolwarm", alpha=0.5)
  plt.grid(True)

def plot_hyperplan(coefs, intercept, legend="hyperplan"):
  xs = np.array([0,1])
  ys = -(coefs[0] * xs + intercept) / coefs[1]
  plt.plot(xs, ys, label=legend)
  plt.legend()

coefs = [1, 1]
intercept = -1
X, y = create_dataset(20, coefs, intercept)

plt.figure(figsize=(5, 5))
plot_dataset(X, y)
plot_hyperplan(coefs, intercept)
plt.show()
```

```python
from abc import ABC, abstractmethod

class Optimizer(ABC):
  @abstractmethod
  def get_w(self, X, y):
    pass

def include_bias(X):
  return np.c_[np.ones((X.shape[0],1)), X]

class SGD(Optimizer):
  def __init__(self, max_iter=1000, learning_rate=0.05):
    self.max_iter = max_iter
    self.learning_rate = learning_rate

  def get_w(self, X, y):
    Xb = include_bias(X)
    y = y.reshape(-1)
    w = np.random.rand(Xb.shape[1])
    for _ in range(self.max_iter):
      errors = y - Xb @ w
      w += self.learning_rate * Xb.T @ errors
    return w

class PseudoInverse(Optimizer):
  def get_w(self, X, y):
    Xb = include_bias(X)
    y = y.reshape(-1)
    w = np.linalg.pinv(Xb) @ y
    return w
```

```python
from sklearn.base import BaseEstimator, ClassifierMixin
from sklearn.metrics import accuracy_score

class ShallowNN(BaseEstimator, ClassifierMixin):
  def __init__(self, optimizer=PseudoInverse()):
    self.optimizer = optimizer

  def fit(self, X, y):
    Xb = include_bias(X)
    y = y.reshape(-1)
    self.w = self.optimizer.get_w(X, y)
    self.intercept = self.w[0]
    self.coefs = self.w[1:]
    return self

  def pre_activation(self, X):
    Xb = include_bias(X)
    logits = Xb @ self.w
    return logits

  def predict(self, X):
    logits = self.pre_activation(X)
    y_pred = np.sign(logits)
    return y_pred

clf = ShallowNN()
clf.fit(X, y)
y_pred = clf.predict(X)
print(f"accuracy: {accuracy_score(y, y_pred)}")
print(f"coefs: {clf.coefs}")

plt.figure(figsize=(5, 5))
plot_dataset(X, y)
plot_hyperplan(coefs, intercept, "generator")
plot_hyperplan(clf.coefs, clf.intercept, "ShallowNN")
plt.show()
```

```python
clf = ShallowNN(SGD(1000,1e-2))
clf.fit(X, y)
y_pred = clf.predict(X)
print(f"accuracy: {accuracy_score(y, y_pred)}")
print(f"coefs: {clf.coefs}")

plt.figure(figsize=(5, 5))
plot_dataset(X, y)
plot_hyperplan(coefs, intercept, "generator")
plot_hyperplan(clf.coefs, clf.intercept, "ShallowNN")
plt.show()
```
